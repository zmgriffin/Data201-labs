8 f) Create 'Elite' variable (qualitative) by binning the Top10per into two groups --> yes if proportion of students coming from the top ten% of their class is over 50%

Top10perc values aren't decimals, so divide by 100 to make them decimals.

```python
college['Elite'] = pd.cut(college['Top10perc']/100,
                          [0,0.5,1],
                          labels= ['No', 'Yes'])

college['Elite'].value_counts()
```

Boxplots of Outstate vs Elite:

```python
fig, ax = plt.subplots(figsize =(8, 6));
college.boxplot('Outstate', by='Elite', ax=ax);
ax.set_xlabel("Elite");
ax.set_ylabel("Out of State Tuition $");
ax.set_title("Out of State Tutition by Elite Status");
```
